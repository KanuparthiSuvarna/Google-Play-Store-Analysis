```python
def convert_size(size):
    if size == 'Varies with device':
        return None
    elif size.endswith('M'):
        return float(size.replace('M', ''))
    elif size.endswith('k'):
        return float(size.replace('k', '')) / 1000
    return None

df['Size'] = df['Size'].apply(convert_size)
```

```python
import pandas as pd

df_reviews = pd.read_csv("googleplaystore_user_reviews.csv")

df_reviews.head()
```

```python
df_reviews.shape
```

```python
df_reviews.isnull().sum()
```

```python
df_reviews = df_reviews.dropna(subset=['Translated_Review'])
```

```python
df_reviews.shape

```

```python
df_reviews.isnull().sum()
```

```python
df_reviews.duplicated().sum()
```

```python
df_reviews = df_reviews.drop_duplicates().reset_index(drop=True)
```

```python
df_reviews.shape
```

```python
df_reviews.duplicated().sum()
```

```python
df_reviews['Sentiment'].value_counts()
```

```python
sentiment_counts = df_reviews['Sentiment'].value_counts()

sentiment_counts.plot(
    kind='bar',
    figsize=(8, 5)
)

plt.title('Distribution of App Reviews by Sentiment')
plt.xlabel('Sentiment')
plt.ylabel('Number of Reviews')
plt.xticks(rotation=0)
plt.show()
```

```python
positive_reviews = df_reviews[
    df_reviews['Sentiment'] == 'Positive'
]

top_positive_apps = positive_reviews['App'].value_counts().head(10)

print(top_positive_apps)
```

```python
negative_reviews = df_reviews[
    df_reviews['Sentiment'] == 'Negative'
]

top_negative_apps = negative_reviews['App'].value_counts().head(10)

print(top_negative_apps)
```

```python
top_negative_apps.plot(
    kind='bar',
    figsize=(10, 5),
    color='salmon'
)

plt.title('Top 10 Apps with Negative Reviews')
plt.xlabel('App Name')
plt.ylabel('Number of Negative Reviews')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
sentiment_counts = df_reviews['Sentiment'].value_counts()

sentiment_counts.plot(
    kind='bar',
    figsize=(8, 5)
)

plt.title('Comparison of App Review Sentiments')
plt.xlabel('Sentiment')
plt.ylabel('Number of Reviews')
plt.xticks(rotation=0)
plt.show()
```

```python
sentiment_percentage = (
    df_reviews['Sentiment'].value_counts(normalize=True) * 100
)

print(sentiment_percentage.round(2))
```

```python
df.to_csv("cleaned_googleplaystore.csv", index=False)

df_reviews.to_csv("cleaned_googleplaystore_reviews.csv", index=False)

print("Both cleaned datasets saved successfully!")
```
